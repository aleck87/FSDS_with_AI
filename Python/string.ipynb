```python
#### strings are immmutable ---- why ?

#they are the fixed values in nature of python, thought the variable holding the value of string can be made to point out another string !
#thats why 
str = "hallo " #pointing to location 0xABC
hex(id(str))
str1 = "world" #pointing to location 0xXYZ
hex(id(str1))

str[0] = "H"   #it wont work bcaz you are trying to change the fixed value 
str = str + " " + str1   # it will work bcaz now str is point to completely different location which is holding a new string value
hex(id(str))
```

```python
#### strings are immmutable ---- why ?

#they are the fixed values in nature of python, thought the variable holding the value of string can be made to point out another string !
#thats why 
str = "hallo " #pointing to location 0xABC
print(str)
print(hex(id(str)))
str1 = "world" #pointing to location 0xXYZ
print(str1)
print(hex(id(str1)))

#str[0] = "H"   #it wont work bcaz you are trying to change the fixed value 
str = str + " " + str1   # it will work bcaz now str is point to completely different location which is holding a new string value
print(str)
print(hex(id(str)))
```
